```python
import matplotlib
matplotlib.use('Agg')
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
import os

# Create the "cfos_plots" subfolder if it doesn't exist
plot_dir = "cfos_plots_behavior_recentSOutlierRearing"
if not os.path.exists(plot_dir):
    os.makedirs(plot_dir)

# Load the data
df = pd.read_csv("behavioral_data_recentSOutlierRearing.csv")

# Separate the groups
vehicle_group = df[df['Treatment'] == 'vei'].drop(['Subject', 'Treatment'], axis=1)
inh_all_group = df[df['Treatment'] == 'mif'].drop(['Subject', 'Treatment'], axis=1)

# Function to calculate Spearman correlation matrix and p-values
def spearman_corr(data):
    corr_matrix = data.corr(method='spearman')
    p_values = data.corr(method=lambda x, y: stats.spearmanr(x, y)[1])
    return corr_matrix, p_values

# Calculate correlations for each group
vehicle_corr, vehicle_p = spearman_corr(vehicle_group)
inh_all_corr, inh_all_p = spearman_corr(inh_all_group)

# save CSV files for correlation matrices and p-values
vehicle_corr.to_csv(os.path.join(plot_dir, "vehicle_corr_matrix.csv"))
vehicle_p.to_csv(os.path.join(plot_dir, "vehicle_p_values.csv"))
inh_all_corr.to_csv(os.path.join(plot_dir, "inh_all_corr_matrix.csv"))
inh_all_p.to_csv(os.path.join(plot_dir, "inh_all_p_values.csv"))

# Function to create annotation text with asterisks for significant correlations
def create_annot(corr_matrix, p_values):
    annot = corr_matrix.round(2).astype(str)
    for i in range(len(corr_matrix)):
        for j in range(len(corr_matrix)):
            if p_values.iloc[i, j] < 0.05:
                annot.iloc[i, j] = annot.iloc[i, j] + '*'
    return annot

# Function to plot and save correlation matrix
def plot_corr_matrix(corr_matrix, p_values, title, filename):
    plt.figure(figsize=(12, 10))
    
    # Mask the upper triangle
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    
    # Create annotation with asterisks
    annot = create_annot(corr_matrix, p_values)
    
    sns.heatmap(corr_matrix, mask=mask, cmap='coolwarm', vmin=-1, vmax=1, center=0,
                square=True, linewidths=.5, cbar_kws={"shrink": .5}, annot=annot, fmt='')
    plt.title(title)
    plt.tight_layout()
    
    # Save plots in the "cfos_plots" subfolder
    plt.savefig(os.path.join(plot_dir, f"{filename}.png"), dpi=300)
    plt.savefig(os.path.join(plot_dir, f"{filename}.eps"), format='eps')
    plt.close()

# Plot and save correlation matrices
plot_corr_matrix(vehicle_corr, vehicle_p, "Vehicle Group Correlation Matrix", "vehicle_corr_matrix")
plot_corr_matrix(inh_all_corr, inh_all_p, "Mifepristone Group Correlation Matrix", "mif_corr_matrix")

print(f"Plots have been saved in the '{plot_dir}' folder.")
```
